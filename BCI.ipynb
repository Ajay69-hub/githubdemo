```python
import torch
from torchvision import datasets, transforms
import numpy as np
```

```python
# define a transform to convert the imgae to tensor
transform = transforms.Compose([transforms.ToTensor()])
```

```python
#define the minist training datasets
mnist_train = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
# download the mnist test datasets
mnist_train = datasets.MNIST(root='./data', train=False, download=True, transform=transform)
```

```python
# Convert to NumPy arrays
train_data_np = mnist_train.data.numpy()
```

```python
image = train_data_np[0]
```
